```python
# Estimating noise

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
import numpy as np
import pandas as pd

df = pd.read_csv('approach_3_C+P.csv')

list_3 = ['Time to Maturity','Risk-Free Rate','Moneyness','mean_moneyness','previous_option_price_scaled',
          'mean_log_return_open','mean_log_return_high','mean_log_return_low','mean_log_return_close',
          'sqrt_var_open','sqrt_var_high','sqrt_var_low', 'sqrt_var_close', 'sqrt_cov_open_high',
          'sqrt_cov_open_low', 'sqrt_cov_open_close', 'sqrt_cov_high_low','sqrt_cov_high_close', 'sqrt_cov_low_close', 
          'Bin']

X_real = df[list_3]
y_real = df['Bin']

model = RandomForestRegressor(random_state=42)
model.fit(X_real, y_real)
y_pred = model.predict(X_real)
residuals = y_real - y_pred
noise_estimate = np.std(residuals)  # or np.var(residuals)
print(noise_estimate)
```

```python
# Estimating redundant_features

import pandas as pd

# Assuming X_real is your feature matrix
corr_matrix = pd.DataFrame(X_real).corr()
import numpy as np

# Get upper triangle of correlation matrix
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Find features with correlation above a threshold
correlation_threshold = 0.99
high_corr_pairs = [(i, j) for i in upper.columns for j in upper.index if abs(upper.loc[j, i]) > correlation_threshold]
redundant_features = set([j for (i, j) in high_corr_pairs])
n_redundant = len(redundant_features)
print(n_redundant)
```
